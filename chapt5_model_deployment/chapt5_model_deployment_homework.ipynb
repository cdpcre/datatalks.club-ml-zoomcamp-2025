````python
import pickle

# 1. Load the pipeline
with open('pipeline_v1.bin', 'rb') as f:
    model = pickle.load(f)

# 2. Input record
record = {
    "lead_source": "paid_ads",
    "number_of_courses_viewed": 2,
    "annual_income": 79276.0
}

# 3. Convert record to list of dicts and predict
X = [record]
y_pred = model.predict_proba(X)[0, 1]

print("Probability that this lead will convert:", y_pred)
````
